```python
# Checking for missing values

missing = df.isnull().sum().sort_values(ascending=False)
missing = missing[missing > 0]

missing_percent = (missing / len(df)) * 100

missing_percent
```

```python
# Visualizing the target variable distribution

import matplotlib.pyplot as plt

df["SalePrice"].hist(bins=50)
plt.title("House Price Distribution")
plt.show()
```

```python
# Visualizing the log-transformed target variable distribution

import numpy as np

np.log1p(df["SalePrice"]).hist(bins=50)
plt.title("Log SalePrice Distribution")
plt.show()
```

```python
# Top 15 features correlated with SalePrice

corr = df.corr(numeric_only=True)

corr["SalePrice"].sort_values(ascending=False).head(15)
```

```python
# Correlation heatmap of top features

import seaborn as sns

plt.figure(figsize=(12,8))
sns.heatmap(corr, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
top_features = corr["SalePrice"].abs().sort_values(ascending=False).head(20)

top_features_df = pd.DataFrame({
    'correlation': top_features,
    'missing_percent': missing_percent.reindex(top_features.index).fillna(0)
})

top_features_df
```
